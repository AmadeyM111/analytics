```python
#if need count in days, not timestamp, then

orders['date'] = orders['order_date'].dt.date
```

```python
result = (
    orders
    .groupby('order_date')
    .agg(orders_count=('order_id', 'count'))
    .sort_values('orders_count', ascending=False)
    .head(1)
)

result
```
