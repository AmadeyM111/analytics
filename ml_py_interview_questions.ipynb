```python
import pandas as pd

orders = pd.DataFrame({
    "order_id": [1,2,3,4,5,6],
    "user_id": [10,10,11,12,11,10],
    "amount": [100,200,150,300,120,80],
    "city": ["Moscow","Moscow","SPB","Kazan","SPB","Moscow"]
})

result = (
    orders
    .groupby("city")
    .agg(
        orders_cnt=("order_id", "count"),
        total_amount=("amount", "sum"),
        avg_amount=("amount", "mean")
    )
    .reset_index()
    .sort_values("total_amount", ascending=False)
)

result
```

```python
users = pd.DataFrame({
    "user_id": [10,11,12],
    "segment": ["premium", "mass", "premium"]
})

result = orders.merge(users, on="user_id", how="left")
result
```

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import f1_score

data = load_breast_cancer()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=43
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

pred = model.predict(X_test)

f1_score(y_test, pred)
```

```python
class MyIterator:
    def __init__(self, data):
        self.index = 0
        self.data = data

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration
        result = self.data[self.index]
        self.index += 1
        return result
```

```python
# decorator that hendle errors and repeat function N times

import functools
def retry(func):
    @functionals.wraps(func)
    def wrapper(*args, **kwargs):
        max_retries = 3
        for i in range(max_retries):
            try:
                result = func(*args, **kwargs)
                return result
            except Exceptions as e:
                print(f'Error occured: {e}. Retrying ({i+1}/{max_retries})...')
            raise Exception(f'Function {func.__name__} failed after {max_retries} attempts.')
    return wrapper

    @retry 
    def my_function(arg1, arg2):
        pass
```
