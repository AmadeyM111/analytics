## Есть таблица заказов:

```python
import pandas as pd

orders = pd.DataFrame({
    'order_id': [1,2,3,4,5,6,7],
    'user_id': [101,102,101,103,102,101,104],
    'amount': [100,200,150,300,250,120,400],
    'order_date': [
        '2025-05-01',
        '2025-05-01',
        '2025-05-02',
        '2025-05-02',
        '2025-05-03',
        '2025-05-03',
        '2025-05-03'
    ]
})
```

## Найти дату с максимальной суммой продаж

```python
result = (
    orders
    .groupby('order_date')
    .agg(total_sales=('amount', 'sum'))
    .sort_values('total_sales', asceding=False)
    .head(1)
)

result
```
